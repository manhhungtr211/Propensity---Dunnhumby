# 04. Temporal Household Behavior

## Mục tiêu

> Hiểu hành vi household theo thời gian: mức độ hoạt động, độ dài lịch sử và khoảng quay lại.

Notebook dùng `DAY` và `WEEK_NO` là ordinal time. Mọi kết luận hiện dựa trên DAY 1–711, chưa gắn được ngày lịch thực.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

import polars as pl

from propensity.eda import (
    BehaviorShiftSpec,
    PurchaseTargetSpec,
    build_household_interpurchase_features,
    build_household_behavior_shift,
    build_household_lifecycle,
    build_household_rolling_windows,
    build_feature_target_associations,
    build_purchase_feature_panel,
    profile_return_intervals,
    profile_historical_activity_coverage,
    profile_window_candidates_across_cutoffs,
    summarize_household_distributions,
    summarize_period_activity,
)
from propensity.eda.dunnhumby import DUNNHUMBY_TRANSACTION_PROFILE_CONFIG
from propensity.eda.visualization import (
    plot_household_interpurchase_dynamics,
    plot_household_behavior_change,
    plot_historical_activity_coverage,
    plot_household_behavior_shift,
    plot_window_candidate_evidence,
    plot_last_purchase_recency,
    plot_lifecycle_coverage_exposure,
    plot_return_interval_distribution,
)
from propensity.io import scan_csv
from propensity.schema.dunnhumby import TRANSACTION_SCHEMA_OVERRIDES

pl.Config.set_tbl_rows(30)
pl.Config.set_tbl_cols(20)
```

## 1. Load transaction data

```python
project_root = Path.cwd().resolve()
if project_root.name == "notebooks":
    project_root = project_root.parent

transactions = scan_csv(
    project_root / "data" / "raw" / "Dunnhumby" / "transaction_data.csv",
    schema_overrides=TRANSACTION_SCHEMA_OVERRIDES,
)
config = DUNNHUMBY_TRANSACTION_PROFILE_CONFIG
```

## 2. Tạo temporal evidence tables

```python
daily_activity = summarize_period_activity(transactions, config.time_col, config)
weekly_activity = summarize_period_activity(
    transactions, config.period_col, config
)
household_distributions = summarize_household_distributions(
    transactions, config
)
household_lifecycle = build_household_lifecycle(transactions, config)
household_interpurchase = build_household_interpurchase_features(
    transactions, config
)
household_windows = build_household_rolling_windows(
    transactions, config, window_weeks=13, step_weeks=4
)
activity_coverage = profile_historical_activity_coverage(
    transactions,
    config,
    windows_days=(3, 7, 14, 21, 30, 60, 90, 180, 365),
)
household_behavior_shift = build_household_behavior_shift(
    transactions,
    config,
    BehaviorShiftSpec(
        recent_days=30, history_days=90, cutoff_step_days=7
    ),
)
purchase_feature_panel = build_purchase_feature_panel(
    transactions,
    config,
    PurchaseTargetSpec(
        observation_days=30,
        horizon_days=7,
        first_cutoff_day=31,
        cutoff_step_days=14,
    ),
)
purchase_target_panel = purchase_feature_panel
feature_target_associations = build_feature_target_associations(
    purchase_feature_panel, min_samples=300
)
window_candidate_evidence = profile_window_candidates_across_cutoffs(
    transactions,
    config,
    windows_days=(7, 14, 30, 60, 90, 180),
    target_horizon_days=30,
    cutoff_step_days=7,
)
return_intervals = profile_return_intervals(transactions, config)
```

## 11. Target thử nghiệm: mua trong 7 ngày tới

Mỗi row là một `(household, cutoff)` đã xuất hiện trước hoặc tại cutoff. Dataset mới kết hợp các feature tổng hợp trực tiếp từ raw transactions trong trailing 30 ngày `[cutoff-29, cutoff]` với target mua bất kỳ sản phẩm nào trong `(cutoff, cutoff+7]`. Cutoff bắt đầu DAY 31, cách nhau 14 ngày và chỉ giữ các target window quan sát đầy đủ.

```python
target_by_cutoff = (
    purchase_target_panel.group_by("cutoff_day")
    .agg(
        pl.len().alias("n_eligible_households"),
        pl.col("target").mean().alias("positive_rate"),
        pl.col("has_full_observation_window").mean().alias(
            "full_exposure_share"
        ),
    )
    .sort("cutoff_day")
)
target_summary = target_by_cutoff.select(
    pl.len().alias("n_cutoffs"),
    pl.col("cutoff_day").min().alias("first_cutoff"),
    pl.col("cutoff_day").max().alias("last_cutoff"),
    pl.col("positive_rate").median().alias("median_positive_rate"),
    pl.col("positive_rate").quantile(0.1).alias("p10_positive_rate"),
    pl.col("positive_rate").quantile(0.9).alias("p90_positive_rate"),
    pl.col("n_eligible_households").min().alias("min_cohort_size"),
    pl.col("n_eligible_households").max().alias("max_cohort_size"),
)
target_summary
```

### Quy ước target

- Không gán target cho household chưa xuất hiện tại cutoff; cohort tăng dần theo first purchase day.
- Giao dịch đúng cutoff chỉ thuộc feature window, không thuộc target. Target bắt đầu từ `cutoff + 1` để tránh leakage.
- Purchase yêu cầu quantity dương và sales không âm; zero-value purchase vẫn hợp lệ, còn return/negative line không tạo positive target.
- `has_full_observation_window` cho biết household đã có đủ 30 ngày exposure hay chưa; không dùng target tương lai để lọc cohort.
- Trên dữ liệu thật có 49 cutoff và 114.685 household-cutoff rows. Median positive rate là 53,3% (P10–P90: 51,5–55,3%). Cohort tăng từ 538 lên 2.500 household.
- Dataset đã được materialize tại `data/processed/purchase_propensity_t30_h7_step14/`; file association giữ correlation theo cutoff và median/P10–P90 để vẽ độ ổn định.

## 14. Phần temporal tiếp theo

Bảng household-level hiện đã có first/last purchase, exposure và basket count. Các bước tiếp theo sẽ bổ sung recency, regularity và continuity trước khi phân tích trend và seasonality. Mọi feature phải chỉ dùng dữ liệu có `DAY <= snapshot` để tránh leakage.
